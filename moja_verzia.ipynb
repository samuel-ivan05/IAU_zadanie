**Zaver:** data je nutne integrovat spojenim `readings` → `units` cez `id_unit` (po zjednoteni formatu) a `units` → `gateways` cez `id_depot`; samotne `id_unit` v `gateways` sa na join neda spoliehat.

### 1.1 B – Analyza jednotlivych atributov (1b)
Pre **min. 10 vyznamnych atributov** analyzujte distribucie a zakladne deskriptivne statistiky (priemer, median, rozptyl, min/max, chybajuce hodnoty...). Overte, ci hodnoty splnaju ocakavany/predpisany rozsah (napr. teploty, tlaky, percenta).

```python
numeric_attrs = {
    "currentMotor": readings["currentMotor"],
    "filterDp": readings["filterDp"],
    "oilTemperature": readings["oilTemperature"],
    "reservoir_pressure": readings["reservoir_pressure"],
    "rms_vibration": readings["rms_vibration"],
    "Flowmeter": readings["Flowmeter"],
    "dewpointDischarge": readings["dewpointDischarge"],
    "serviceHours": units["serviceHours"],
}

# Predpokladane pripustne rozsahy pre kompresor
expected_ranges = {
    "currentMotor":       (0, 20),     # prevadzkovy prud motora
    "filterDp":           (0, 60),     # tlakova strata na filtri
    "oilTemperature":     (-20, 110),  # [°C]
    "reservoir_pressure": (0, 12),     # [bar]
    "rms_vibration":      (0, 15),     # vibracie - musia byt kladne
    "Flowmeter":          (0, 100),    # prietok - musi byt kladny
    "dewpointDischarge":  (-40, 40),   # [°C] rosny bod, zaporne hodnoty su v poriadku
    "serviceHours":       (0, 60000),  # [h] kumulativne najazdene hodiny, nezaporne
}

rows = []
for name, s in numeric_attrs.items():
    lo, hi = expected_ranges[name]
    valid = s.dropna()
    out_of_range = ((valid < lo) | (valid > hi)).sum()
    rows.append({
        "atribút": name,
        "n": s.shape[0],
        "chýbajúce": s.isna().sum(),
        "chýbajúce_%": round(100 * s.isna().mean(), 2),
        "priemer": round(valid.mean(), 3),
        "medián": round(valid.median(), 3),
        "rozptyl": round(valid.var(), 3),
        "smer_odch": round(valid.std(), 3),
        "min": round(valid.min(), 3),
        "max": round(valid.max(), 3),
        "očak._rozsah": f"[{lo}, {hi}]",
        "mimo_rozsahu_n": int(out_of_range),
        "mimo_rozsahu_%": round(100 * out_of_range / len(valid), 2),
    })

numeric_summary = pd.DataFrame(rows).set_index("atribút")
display(numeric_summary)

print("Atribúty s hodnotami mimo očakávaného rozsahu:")
print(numeric_summary.loc[numeric_summary["mimo_rozsahu_n"] > 0, ["min", "max", "očak._rozsah", "mimo_rozsahu_n", "mimo_rozsahu_%"]])
```

```python
# --- 1.1 B: Frekvencne tabulky pre kategoricke/binarne atributy ---

categorical_attrs = {
    "failure": readings["failure"],       # cielova premenna (0/1)
    "class_duty": units["class_duty"],
    "codeModel": units["codeModel"],
    "fleetCode": units["fleetCode"],
}

for name, s in categorical_attrs.items():
    print(f"=== {name} ===")
    print(f"Počet unikátnych hodnôt: {s.nunique(dropna=True)}   "
          f"Chýbajúce: {s.isna().sum()} ({100 * s.isna().mean():.2f} %)")
    vc = s.value_counts(dropna=False)
    display(vc.to_frame("počet").assign(percento=lambda d: (100 * d["počet"] / len(s)).round(2)))
    print()
```

```python
# --- 1.1 B: Vizualizacia distribucii (histogramy s priemerom/medianom a ocakavanym rozsahom, boxploty, bar charty) ---

MEAN_COLOR = "#2563eb"    # modra – priemer
MEDIAN_COLOR = "#f59e0b"  # jantarova – median
RANGE_COLOR = "#dc2626"   # cervena – hranice ocakavaneho rozsahu

fig, axes = plt.subplots(4, 2, figsize=(12, 14))
for ax, (name, s) in zip(axes.ravel(), numeric_attrs.items()):
    valid = s.dropna()
    sns.histplot(valid, kde=True, ax=ax, bins=40, color="#94a3b8")
    ax.axvline(valid.mean(), color=MEAN_COLOR, linestyle="-", linewidth=1.5)
    ax.axvline(valid.median(), color=MEDIAN_COLOR, linestyle="--", linewidth=1.5)
    lo, hi = expected_ranges[name]
    ax.axvline(lo, color=RANGE_COLOR, linestyle=":", linewidth=1)
    ax.axvline(hi, color=RANGE_COLOR, linestyle=":", linewidth=1)
    ax.set_title(name)

legend_handles = [
    plt.Line2D([0], [0], color=MEAN_COLOR, linestyle="-", linewidth=1.5, label="priemer"),
    plt.Line2D([0], [0], color=MEDIAN_COLOR, linestyle="--", linewidth=1.5, label="medián"),
    plt.Line2D([0], [0], color=RANGE_COLOR, linestyle=":", linewidth=1, label="očakávaný rozsah"),
]
fig.legend(handles=legend_handles, loc="upper center", ncol=3, bbox_to_anchor=(0.5, 1.02), frameon=False)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(4, 2, figsize=(12, 14))
for ax, (name, s) in zip(axes.ravel(), numeric_attrs.items()):
    valid = s.dropna()
    sns.boxplot(
        x=valid, ax=ax, color="#94a3b8",
        showmeans=True,
        meanprops={"marker": "D", "markerfacecolor": MEAN_COLOR, "markeredgecolor": MEAN_COLOR, "markersize": 6},
    )
    ax.set_title(name)

legend_handles = [
    plt.Line2D([0], [0], marker="D", color="none", markerfacecolor=MEAN_COLOR, markeredgecolor=MEAN_COLOR, markersize=6, label="priemer"),
    plt.Line2D([0], [0], color="black", linewidth=1.5, label="medián (čiara v boxe)"),
]
fig.legend(handles=legend_handles, loc="upper center", ncol=2, bbox_to_anchor=(0.5, 1.02), frameon=False)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, (name, s) in zip(axes.ravel(), categorical_attrs.items()):
    s.value_counts(dropna=False).plot(kind="bar", ax=ax, color="#94a3b8")
    ax.set_title(name)
    ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

**Komentar k distribuciam a deskriptivnym statistikam:**

*Ciselne atributy (tabulka + histogramy + boxploty):*
- **`reservoir_pressure`, `Flowmeter`, `dewpointDischarge`** – priemer ≈ median, nizky rozptyl vzhladom na rozsah, symetricky "zvonovy" tvar, 0 hodnot mimo ocakavaneho rozsahu. Ide o najcistejsie atributy.
- **`currentMotor`** – priemer (5.33) mierne nad medianom (5.19) → jemne zosikmenie doprava. 11 hodnot (0.07 %) je mimo [0, 20], pricom max = 99 je extremna odlahla hodnota (fyzicky nerealny prud) – kandidat na outlier/chybu.
- **`filterDp`** – priemer (24.13) citelne nad medianom (21.43) → pravostranne zosikmenie, potvrdene aj na histograme/boxplote dlhym chvostom. 606 chybajucich (4.06 %) a 203 hodnot (1.42 %) mimo [0, 60] (max = 120).
- **`oilTemperature`** – priemer (51.64) je *nizsi* ako median (54.95), co je nezvycajne opacnym smerom – sposobuje to 47 hodnot rovnych **-999**, co nie je realna teplota, ale typicky **sentinelovy/chybovy kod pre chybajucu hodnotu**. Treba ich v 1.2 (A/B) prekodovat na NaN, nie liecit ako bezny outlier.
- **`rms_vibration`** – priemer (3.06) nad medianom (2.72) → pravostranne zosikmenie, 539 chybajucich (3.61 %), ale vsetky hodnoty v ramci fyzikalneho rozsahu.
- **`serviceHours`** – velky rozptyl (smer. odch. ≈ 8928 pri priemere ≈ 14313) a priemer zretelne nad medianom (14313 vs. 12014) → flotila jednotiek ma velmi roznorody "vek"/najazd, s pravostrannym chvostom smerom k starsim jednotkam; 100 chybajucich (5 %).

*Kategoricke atributy (frekvencne tabulky + bar charty):*
- **`failure`** – silne nevyvazena cielova premenna (87.99 % / 12.01 %) → treba zohladnit pri volbe metrik a modelovania vo Faze 3.
- **`class_duty`** – formalne 20 unikatnych hodnot, v skutocnosti ide len o **~5 realnych kategorii** (napr. `light`/`L`/`1`) zapisanych nekonzistentne (rozny casing, skratky vs. plne nazvy, medzera na zaciatku ako `" L"`) → problem s formatom na zjednotenie v 1.2 A, nie skutocna vysoka kardinalita.
- **`codeModel`** a **`fleetCode`** – ciste, bez chybajucich hodnot, relativne vyvazene kategorie (kazda cca 18–26 %).

**Zhrnutie:** Rozdiel priemer vs. median spolu s bodkami za fuzmi na boxplotoch signalizuje pravostranne zosikmenie/potencialne outliery hlavne pri `currentMotor`, `filterDp`, `rms_vibration` a `serviceHours` – tie riesime statisticky (IQR/percentily) v **1.2 C**. `oilTemperature` (-999) a hodnoty mimo `expected_ranges` pri `currentMotor`/`filterDp` su skor **domenovo/fyzikalne nemozne hodnoty** – tie riesime ako chyby v **1.2 B**. `class_duty` je problem nejednotneho formatu, nie outlierov – riesi sa v **1.2 A**.
